## Part 2: The Power Question

<div style="font-size:1.5em;">
  <p><strong>Quick fire:</strong> answer without Python.</p>
  <div class="qa-stack">
    <div class="qa-step fragment fade-out" data-fragment-index="0">
      <p><strong>10 to the power of <em>what</em> gives 100?</strong></p>
      <p style="font-size:0.8em;">Think: how many times do you multiply 10 by itself to get 100?</p>
    </div>
    <div class="qa-step fragment fade-in-then-out" data-fragment-index="0">
      <p style="color:#2a7a2a;">That&rsquo;s right! <strong>2</strong>, because 10 &times; 10 = 10<sup>2</sup> = 100.</p>
      <p><strong>10 to the power of <em>what</em> gives 10?</strong></p>
    </div>
    <div class="qa-step fragment fade-in-then-out" data-fragment-index="1">
      <p style="color:#2a7a2a;">That&rsquo;s right! <strong>1</strong>, because 10<sup>1</sup> is just 10 itself.</p>
      <p><strong>10 to the power of <em>what</em> gives 1?</strong></p>
      <p style="font-size:0.8em;">(Tricky! What happens when the power is 0?)</p>
    </div>
    <div class="qa-step fragment fade-in-then-out" data-fragment-index="2">
      <p style="color:#2a7a2a;">That&rsquo;s right! <strong>0</strong>, because any number to the power of 0 is 1. Try <code>10**0</code> in Python &mdash; you get 1.</p>
      <p><strong>So what is log(1)?</strong></p>
    </div>
    <div class="qa-step fragment" data-fragment-index="3">
      <p style="color:#2a7a2a;">That&rsquo;s right! <strong>log(1) = 0</strong>.</p>
      <p>Key fact: <strong>log of 1 is always 0</strong>, no matter what the base is. Because anything<sup>0</sup> = 1.</p>
    </div>
  </div>
</div>

<pre><code>def my_log(value, base, lo, hi, iterations):

    &mdash; now packaged as a reusable function.</p>

## Part 6: Beyond Base 10

Up to now, every &ldquo;log&rdquo; you computed was secretly <strong>log base 10</strong> &mdash;
counting zeros only works because our number system is built on powers of 10.
From here on we&rsquo;ll write the base explicitly, like log<sub>2</sub> or log<sub>3</sub>,
because the base is about to change.


<div style="font-size:1.5em;">
  <p>There&rsquo;s nothing special about 10. Let&rsquo;s try <strong>base 2</strong>.</p>
  <p>log<sub>2</sub>(x) asks: <strong>2 to the power of <em>what</em> gives x?</strong></p>

## Part 7: The Change of Base Formula

<div style="font-size:1.5em;">
  <p>Python gives you <code>math.log10()</code> and <code>math.log2()</code> built in.
  But what if you need log base 3? Or base 5?</p>
  <div class="fragment" data-fragment-index="0">
    <p>Try this experiment. For each row, <strong>divide log<sub>10</sub>(value) by log<sub>10</sub>(base)</strong>:</p>
    <table style="margin:0.5em auto; font-size:0.85em;">
      <tr><th style="padding:0.2em 0.6em;">Want</th><th style="padding:0.2em 0.6em;">log<sub>10</sub>(value)</th><th style="padding:0.2em 0.6em;">log<sub>10</sub>(base)</th><th style="padding:0.2em 0.6em;">ratio</th><th style="padding:0.2em 0.6em;">expected</th></tr>
      <tr><td>log<sub>3</sub>(27)</td><td>1.431</td><td>0.477</td><td><span class="fragment fade-out" data-fragment-index="1">?</span><span class="fragment" data-fragment-index="1">3.0</span></td><td>3</td></tr>

<div style="font-size:1.5em;">
  <p>You just discovered the <strong>change of base formula</strong>:</p>
  <p style="text-align:center; font-size:1.3em; color:#7a6b3a; margin:0.5em 0;">
    <strong>log<sub>b</sub>(x) = log<sub>10</sub>(x) &divide; log<sub>10</sub>(b)</strong>
  </p>
  <div class="fragment" data-fragment-index="0">
    <p>This means you can compute a log in <em>any</em> base using only log<sub>10</sub>:</p>
<pre><code>import math
def log_base(value, base):
    return math.log10(value) / math.log10(base)</code></pre>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p><strong>Test it:</strong></p>
    <p style="font-size:0.9em;">
    <code>log_base(27, 3)</code> &rarr; 3.0 &nbsp;&#10003; &nbsp;
    <code>log_base(6, 2)</code> &rarr; 2.585 &nbsp;&#10003; &nbsp;
    <code>log_base(1024, 2)</code> &rarr; 10.0 &nbsp;&#10003;</p>
  </div>
  <div class="fragment" data-fragment-index="2">
    <p>The base in the denominator can be anything &mdash;
    log<sub>2</sub>, log<sub>10</sub>, ln &mdash; the ratio is always the same.</p>
    <p style="font-size:0.85em;">(Python&rsquo;s <code>math.log(value, base)</code> does exactly this internally.)</p>
